```python
import altair as alt
import datetime as dt
import numpy as np
import pandas as pd
import requests

from scipy.stats import norm
from toolz.curried import *

alt.data_transformers.disable_max_rows()
```

```python
BASE_URL = "https://thalex.com/api/v2/public"

dfs = {}

timestamps = (
    int(dt.datetime(2026, 2, 6, 8, 0).timestamp()),
    int(dt.datetime(2026, 2, 13, 23, 0).timestamp()),
)

resolution = "1h"
```

```python
dfs["index"] = pipe(
    requests.get(
        f"{BASE_URL}/index_price_historical_data",
        params={
            "index_name": "BTCUSD",
            "resolution": resolution,
            "from": timestamps[0],
            "to": timestamps[1],
        },
    ),
    requests.Response.json,
    get_in(["result", "index"]),
    curry(
        pd.DataFrame,
        columns=[
            "ts",
            "index_price_open",
            "index_price_high",
            "index_price_low",
            "index_price_close",
        ],
    ),
)
# .assign(date_time=lambda df: pd.to_datetime(df["ts"], unit="s"))

dfs["index"]
```

```python
all_instrument_names = requests.get(f"{BASE_URL}/all_instruments").json()["result"]
```

```python
expiry_date = "2026-02-27"

calls = [
    i["instrument_name"]
    for i in all_instrument_names
    if i["product"] == "OBTCUSD"
    and i["option_type"] == "call"
    and i["expiry_date"] == expiry_date
    # and i["strike_price"] >= 115_000
    # and i["strike_price"] <= 140_000
]

puts = [
    i["instrument_name"]
    for i in all_instrument_names
    if i["product"] == "OBTCUSD"
    and i["option_type"] == "put"
    and i["expiry_date"] == expiry_date
    # and i["strike_price"] >= 80_000
    # and i["strike_price"] <= 115_000
]

instrument_names = calls + puts
instrument_names
```

```python
dfs["mark"] = (
    pipe(
        {
            instrument_name: requests.get(
                f"{BASE_URL}/mark_price_historical_data",
                params={
                    "from": timestamps[0],
                    "to": timestamps[1],
                    "resolution": resolution,
                    "instrument_name": instrument_name,
                },
            )
            for instrument_name in instrument_names
        },
        valmap(requests.Response.json),
        valmap(get_in(["result", "mark"])),
        valmap(
            curry(
                pd.DataFrame,
                columns=[
                    "ts",
                    "mark_price_open",
                    "mark_price_high",
                    "mark_price_low",
                    "mark_price_close",
                    "iv_open",
                    "iv_high",
                    "iv_low",
                    "iv_close",
                    "tob",
                ],
            )
        ),
        pd.concat,
    )
    .droplevel(1)
    .reset_index(names=["instrument_name"])
    .assign(date_time=lambda df: pd.to_datetime(df["ts"], unit="s"))
    .assign(k=lambda df: df["instrument_name"].map(lambda s: int(s.split("-")[2])))
)

dfs["mark"]
```

```python
instrument_lookup = {data["instrument_name"]: data for data in all_instrument_names}
```

```python
def calc_delta(spot, strike, tte, iv, option_type="call"):
    tau = tte / (365.25 * 24 * 60 * 60)
    d1 = (np.log(spot / strike) + 0.5 * iv**2 * tau) / (iv * np.sqrt(tau))
    if option_type == "call":
        return norm.cdf(d1)
    else:
        return norm.cdf(d1) - 1


def calc_vega(spot, strike, tte, iv):
    tau = tte / (365.25 * 24 * 60 * 60)
    d1 = (np.log(spot / strike) + 0.5 * iv**2 * tau) / (iv * np.sqrt(tau))
    return spot * norm.pdf(d1) * np.sqrt(tau)


def calc_gamma(spot, strike, tte, iv) -> float:
    tau = tte / (365.25 * 24 * 60 * 60)
    d1 = (np.log(spot / strike) + 0.5 * iv**2 * tau) / (iv * np.sqrt(tau))
    return norm.pdf(d1) / (spot * iv * np.sqrt(tau))


def calc_theta(spot, strike, tte, iv, option_type):
    tau = tte / (365.25 * 24 * 60 * 60)
    d1 = (np.log(spot / strike) + 0.5 * iv**2 * tau) / (iv * np.sqrt(tau))
    theta_annual = -(spot * norm.pdf(d1) * iv) / (2 * np.sqrt(tau))
    return theta_annual / 365.25


dfs["pnl"] = pd.merge(dfs["index"], dfs["mark"], on="ts").assign(
    date_time=lambda df: pd.to_datetime(df["ts"], unit="s"),
    expiration=lambda df: df["instrument_name"].map(
        lambda name: get_in([name, "expiration_timestamp"], instrument_lookup)
    ),
    option_type=lambda df: df["instrument_name"].map(
        lambda name: get_in([name, "option_type"], instrument_lookup)
    ),
    strike=lambda df: df["instrument_name"]
    .map(lambda name: get_in([name, "strike_price"], instrument_lookup))
    .astype(int),
    tte=lambda df: df["expiration"] - df["ts"],
    delta=lambda df: df.apply(
        lambda row: calc_delta(
            row["index_price_close"],
            row["strike"],
            row["tte"],
            row["iv_close"],
            row["option_type"],
        ),
        axis=1,
    ),
    gamma=lambda df: df.apply(
        lambda row: calc_gamma(
            row["index_price_close"],
            row["strike"],
            row["tte"],
            row["iv_close"],
        ),
        axis=1,
    ),
    theta=lambda df: df.apply(
        lambda row: calc_theta(
            row["index_price_close"],
            row["strike"],
            row["tte"],
            row["iv_close"],
            row["option_type"],
        ),
        axis=1,
    ),
    vega=lambda df: df.apply(
        lambda row: calc_vega(
            row["index_price_close"],
            row["strike"],
            row["tte"],
            row["iv_close"],
        ),
        axis=1,
    ),
    delta_abs=lambda df: abs(df["delta"]),
)

dfs["pnl"]
```
